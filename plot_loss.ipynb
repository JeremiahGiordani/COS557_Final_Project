Then, do this for Epoch 2, 3, ... and put those values into the list below

```python
import matplotlib.pyplot as plt

# Loss values will go here. You can find them by looking at the training logs
# Here, I pasted all of the values for loss in the 15_iterations_training.log file
losses = [
    1.3831, 1.1459, 0.9088, 0.6683, 0.4736, 
    0.4799, 0.3635, 0.2813, 0.1036, 0.3616, 
    0.1574, 0.0967, 0.0607, 0.0422, 0.0317
] 

# X-axis: index + 1 for each value
x = [i + 1 for i in range(len(losses))]

# Plotting the line plot
plt.plot(x, losses, marker='o')
plt.title("Training Loss over Epoch")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.grid(True)
plt.show()
```
